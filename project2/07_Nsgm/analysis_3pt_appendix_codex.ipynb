# Light-current GEVP comparisons on A24 and A48

Display the paired ratios exported by the ensemble analyses. No additional fits or filter choices are made here.

```python
from pathlib import Path
import os, sys

ROOT = Path.cwd().resolve()
if ROOT.name == "__codex_ignore":
    ROOT = ROOT.parent
assert ROOT.name == "07_Nsgm"
WORK = ROOT / "__codex_ignore"
os.chdir(WORK)
sys.path.insert(0, str(ROOT))
import util as yu
import util_codex as yuc

yu.setpath("analysis_3pt_appendix_codex")
datasets = []
for ensemble in ["cA211.530.24", "cA2.09.48"]:
    data = yu.load_pkl(WORK / ensemble / "pkl/analysis_3pt_light_codex/reg_ignore/appendix_ratios.pkl")
    if data is None:
        raise FileNotFoundError(f"Run {ensemble}/analysis_3pt_light_codex.ipynb first.")
    assert data["filter_displacement"] == 2
    assert set(data["ratios"]["standard"]) == set(data["ratios"]["gevp"]) == set(data["ratios"]["rlg"])
    datasets.append(data)
```
