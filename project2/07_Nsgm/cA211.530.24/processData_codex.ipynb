# Appendix D contraction data: cA211.530.24

Process the local HDF5 export into a small pickle, preserving the production blocking, vacuum subtraction, and reconstructed source/sink partners. The full matrices are checked against the existing production pickle before saving. This does not change that pickle or remove the raw file.


```python
from pathlib import Path
import os, sys

HERE = Path.cwd().resolve()
assert HERE.name == "cA211.530.24"
PROJECT = HERE.parent
sys.path.insert(0, str(PROJECT))
import util as yu
from util_codex import compact_topology_data

RAW = Path.home() / "Desktop/NST_f_cA211.53.24_Nsgm.h5"
REFERENCE = HERE / "pkl/processData/reg_ignore/data.pkl"
OUTPUT = PROJECT / "__codex_ignore" / HERE.name
OUTPUT.mkdir(parents=True, exist_ok=True)
os.chdir(OUTPUT)
yu.setpath("processData_codex")
```

## Reconstruct and validate

Read only the zero-momentum scalar channel needed here. Both irrep rows and all production separations are retained. On B64, reuse the connected nucleon contribution already stored in the production pickle. The two truncations change only off-diagonal three-point entries in the later analysis.


```python
topologies = compact_topology_data(RAW, REFERENCE, block_size=4, time_extent=16)
metadata = topologies["metadata"]
print("Production matrices reproduced for", metadata["ensemble"])
print("Separations:", metadata["separations"])
print("Configurations:", len(metadata["configurations"]), "block size:", metadata["block_size"])
print("Existing unused tail:", metadata["discarded_tail"])
```

## Save

The pickle includes full two-/three-point matrices, matched two-point denominators, BWZ and direct contributions, configuration IDs, blocking, and source hashes. It contains no fitted weights; Appendix D reads the current two-point analysis weights separately. After verification, its analysis needs only pickle inputs.


```python
yu.save_pkl_reg("data_topologies", topologies)
path = OUTPUT / "pkl/processData_codex/reg_ignore/data_topologies.pkl"
print(path)
print(f"Size: {path.stat().st_size / 1024**2:.2f} MiB")
```
