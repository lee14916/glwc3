```python
%load_ext autoreload
%autoreload 2

import util as yu
from util import *
import util_moments as yum

yu.setpath('plot_paper')

tftcphy_A20_conn=tftcphy_B20_conn=(0.8,0.3)
tftcphy_A20_discq=tftcphy_B20_discq=(0.6,0.2)
tftcphy_A20_gluon=tftcphy_B20_gluon=(0.6,0.3)
```

```python
ens='e'; j='j-;conn'; symmetrizeQ=True
fits=yu.getFits(label=f'{ens}_{j}_2st_{symmetrizeQ}',pathlabel='analysis_avgx_2')
fitlabels=[fit[0] for fit in fits]
tfmins=list(set([fitlabel[0] for fitlabel in fitlabels])); tfmins.sort()
tcmins=list(set([fitlabel[1] for fitlabel in fitlabels])); tcmins.sort()
tfphy,tcphy=tftcphy_A20_conn
ind=np.argmin([np.abs(tfmin*yu.ens2a[ens] - tfphy) + np.abs(tcmin*yu.ens2a[ens] - tcphy) for tfmin,tcmin in fitlabels])

fig,axs=yu.getFigAxs(1,1,Lrow=4,Lcol=8)
ax=axs[0,0]
xunit=yu.ens2a[ens]; yunit=1

for ifit,fit in enumerate(fits):
    (tfmin,tcmin),pars_jk=fit[:2]
    itcmin=tcmins.index(tcmin)
    if itcmin>7:
        continue
    color=yu.colors8[itcmin]; fmt=yu.fmts8[itcmin]
    
    mean,err=yu.jackme(pars_jk[:,0])
    
    plt_x=(tfmin+itcmin*0.2)*xunit; plt_y=mean*yunit; plt_yerr=err*yunit
    ax.errorbar(plt_x,plt_y,plt_yerr,color=color, fmt=fmt, mfc='white' if ifit==ind else None, label=tcmin if tfmin==tfmins[-1] else None, markersize=10)
    
    if ifit==ind:
        ax.axhspan(plt_y-plt_yerr,plt_y+plt_yerr,color=color,alpha=0.2)

ax.set_ylim([0.115,0.165]) 
# ax.set_yticks([0.05,0.1,0.15,0.2])
ax.legend(loc="upper right", frameon=True, ncol=len(tcmins),
                  fontsize=22, handlelength=0.9, columnspacing=0.45,
                  handletextpad=0.25, borderpad=0.25)

ax.set_xlabel(r'$t_s^{\rm low}$ [fm]')
ax.set_ylabel(r'$A_{20}^{u-d}(0)$')
yu.finalizePlot('rainbow_avgx_2/j-;conn_equal_tins')
```
